```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/survey_data.csv')
df.head()
```

## Cleaning
Remove duplicates, standardize mode labels, convert numeric columns, and fill missing numeric values with medians.

```python
df = df.drop_duplicates()
df['Commute_Mode'] = df['Commute_Mode'].astype(str).str.strip().str.title()
for col in ['Distance_km','Travel_Time_min','Daily_Cost_INR']:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    df[col] = df[col].fillna(df[col].median())
df.head()
```

## Summary statistics

```python
print('Students:', len(df))
print('Average distance (km):', round(df.Distance_km.mean(),2))
print('Average one-way travel time (min):', round(df.Travel_Time_min.mean(),1))
print('Average daily cost (INR):', round(df.Daily_Cost_INR.mean(),2))
print('Most popular mode:', df.Commute_Mode.mode().iloc[0])
df.describe()
```

## Visualizations

```python
sns.countplot(data=df, x='Commute_Mode', order=df.Commute_Mode.value_counts().index)
plt.xticks(rotation=30); plt.title('Students by Commute Mode'); plt.tight_layout()
```

```python
sns.histplot(df.Distance_km, bins=12, kde=True); plt.title('Commute Distance Distribution'); plt.show()
```

```python
sns.scatterplot(data=df, x='Distance_km', y='Travel_Time_min', hue='Commute_Mode'); plt.title('Distance vs Travel Time'); plt.show()
```

## Sustainability extension
Illustrative factors are teaching assumptions only. Consult a current official source and document assumptions for a formal emissions estimate.

```python
factors = {'Bus':0.08,'Car':0.18,'Carpool':0.09,'Motorbike':0.10,'Bicycle':0.0,'Walking':0.0,'Auto-Rickshaw':0.11}
df['Estimated_kg_CO2_oneway'] = df.apply(lambda r: r.Distance_km * factors.get(r.Commute_Mode,0.12), axis=1)
df[['Commute_Mode','Distance_km','Estimated_kg_CO2_oneway']].head()
```
